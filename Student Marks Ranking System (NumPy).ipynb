## Project: Student Marks Ranking System (NumPy)

```python
import numpy as np

# Marks of students (5 students × 3 subjects)
marks = np.array([
    [78, 85, 90],
    [60, 72, 68],
    [88, 91, 94],
    [45, 55, 50],
    [70, 75, 80]
])

students = np.array(["Amit", "Rahul", "Sneha", "Kiran", "Pooja"])
```

**Total Marks (per student)**

```python
total_marks = marks.sum(axis=1)
print("Total Marks:", total_marks)
```

**Percentage Calculation**

```python
percentage = (total_marks / 300) * 100
print("Percentage:", percentage)
```

**Grade System**

```python
grade = np.where(
    percentage >= 80, "A",
    np.where(percentage >= 60, "B",
    np.where(percentage >= 40, "C", "Fail"))
)

print("Grades:", grade)
```

**Ranking System for student**

```python
rank = np.argsort(-percentage) + 1
print("Rank:", rank)
```

**Final Student Report**

```python
for i in range(len(students)):
    print(f"{students[i]} | Total: {total_marks[i]} | %: {percentage[i]:.2f} | Grade: {grade[i]} | Rank: {rank[i]}")
```
